```python
from typing import Callable, Any

def checker(func: Callable[[Any], Any]) -> Callable[[Any], Any]:
    def wrapper(self: Any, *args: Any, **kwargs: Any):
        print(args, kwargs)
        for kwarg in kwargs:
            if kwargs[kwarg] == getattr(self, kwarg):
                return None
        return func(self, *args, **kwargs)
    return wrapper

class Test:
    def __init__(self):
        self.val = 10

    @checker
    def test(self, val = 10):
        print("execute", val)

test = Test()
test.test(val=10)
```

```python
def my_decorator(fn):
    def wrapper(self, *args, **kwargs):
        print(self.class_var)
        return fn(self, *args, **kwargs)
    return wrapper


class MyClass:
    class_var = 123

    @my_decorator
    def test(self):
        print("inside test")


obj = MyClass()
obj.test()
```
