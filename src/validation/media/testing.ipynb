```python
import numpy as np

from CuPyNLO.media.fibers.fiber import FiberInstance as Fiber1
from CuPyNLO.media. fibers.fiber_v2 import FiberInstance as Fiber2

from CuPyNLO.light.DerivedPulses import SechPulse as SechPulse1
from CuPyNLO.light.DerivedPulses_v2 import SechPulse as SechPulse2

from CuPyNLO.interactions.FourWaveMixing.SSFM import SSFM as SSFM1
from CuPyNLO.interactions.FourWaveMixing.SSFM_v2 import SSFM as SSFM2
```

```python
power = 1e4
t0_ps = 28.4e-3
cwl = 835

n = 2**13
length = .15
fibertype = "dudley"

dz = 1e-3
steps = 100
range1 = np.arange(steps)
```

```python
fiber1 = Fiber1()
fiber2 = Fiber2()

pulse1 = SechPulse1(power, t0_ps, cwl)
pulse2 = SechPulse2(power, t0_ps, cwl)
```

```python
fiber1.load_from_db(length, fibertype)
fiber2.load_from_db(length, fibertype)
```

```python
evol1 = SSFM1(USE_SIMPLE_RAMAN=True)
evol2 = SSFM2(use_simple_raman=True)
```

```python
print(1e9 * pulse1.calc_epp())
print(1e9 * pulse2.calc_epp())
```

```python
y1, aw1, at1, pulse_out1 = evol1.propagate(pulse1, fiber1, steps)
print("\n----------------------------\n")
y2, aw2, at2, pulse_out2 = evol2.propagate(pulse2, fiber2, steps)
```

```python
print(set(y1 == y2))
print(aw2)
```

```python
print(fiber1.get_gamma())
print(fiber2.gamma)
```
